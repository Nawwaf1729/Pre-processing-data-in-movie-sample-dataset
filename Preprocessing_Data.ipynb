The following code is to find out null and duplicate values:

```python
import numpy as np

# replace "Null" with NaN
df.replace("Null", np.nan, inplace = True)
df.replace("N/A", np.nan, inplace=True)
#Check missing value
print(df.isnull().sum())
```

There are about 2 rows that have same data in 12 rows. So there are 6 duplicate rows. We will drop the duplicated rows and leave the unique one in the Data Cleaning Section.

## Data Cleaning

In this section we will perform data cleaning by:
1. Remove null values in gross and budget columns, since they are the important columns.
2. Remove duplicate values and leave the unique value.
3. Fill null values in color columns with modes.
4. Fill null values in director_name columns with "Unknown".
5. Fill null values in genres column with "None".
6. Correct the negative values in imdb_score column with the absolute value function.
7. Correct the invalid data in duration column such as the duration higher than 600 or less than 10 and fill it with median.

```python
#drop null in gross and budget
df.dropna(subset=['gross', 'budget'], inplace=True)
```

```python
#remove duplicate value and leave one unique value
df.drop_duplicates(inplace=True)
```

```python
#fill null values in 'color' with modes
df['color'].fillna(df['color'].mode()[0], inplace=True)
```

```python
#fill null values in 'director_name' with "Unknown"
df['director_name'].fillna('Unknown', inplace=True)
```

```python
#Fill null values in 'genres' with 'None'
df['genres'].fillna('None', inplace=True)
```

```python
# Correct the invalid data (negative values) in 'imdb_score' with the absolute value function
df.loc[~df['imdb_score'].between(0, 10), 'imdb_score'] = np.abs(df['imdb_score'])
```

```python
#Correct the invalid data in 'duration'
# 1. First, convert the invalid values to NaN so we can ignore them in our calculation
df.loc[df['duration'] <= 10, 'duration'] = np.nan
df.loc[df['duration'] > 600, 'duration'] = np.nan

# 2. Calculate the median of the 'duration' column, ignoring the NaN values
median_duration = df['duration'].median()
print(f"The median movie duration is: {median_duration} minutes")

# 3. Fill the NaN values with the calculated median
df['duration'].fillna(median_duration, inplace=True)
```

```python
#See the result after data cleaning
print(df.isnull().sum())
```

## Data Transformation

In this section we will perform data transformation by:
1. Standardize the color and country columns.
2. Convert some columns to numeric.
3. Splitting genres column

```python
# Clean the 'color' column: standardize lowercase
df['color'] = df['color'].str.lower()

# Clean the 'country' column: standardize capitalization
df['country'] = df['country'].str.upper()
```

```python
# Convert columns to numeric
numeric_cols = ['duration', 'gross', 'title_year', 'budget', 'imdb_score']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
#Splitting 'genres' columns

# 1. Split the 'genres' column by '|' into a new DataFrame
genre_split_df = df['genres'].str.split('|', expand=True)

# 2. Create new column names like 'genre_1', 'genre_2', etc.
new_column_names = [f'genre_{i+1}' for i in range(genre_split_df.shape[1])]
genre_split_df.columns = new_column_names

# 3. Concatenate the new genre columns with the original DataFrame
df_with_genres = pd.concat([df, genre_split_df], axis=1)
df_with_genres.fillna('None', inplace=True)

# Result
print(df_with_genres.head())
```

```python
#Save the result
df_with_genres.to_csv('/content/movie_dataset_cleaned.csv', index=False)
```
